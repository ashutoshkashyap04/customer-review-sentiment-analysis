```python
from datasets import load_dataset, concatenate_datasets
```

```python
dataset = load_dataset("Yelp/yelp_review_full")
dataset
```

```python
# Filtering 20k reviews from training dataset  for each class => 100k rows in total

train = dataset["train"].shuffle(seed = 42)

train_subset = []

for label in range(5):
    subset = train.filter(
        lambda x: x["label"] == label
    ).select(range(20000))
    
    train_subset.append(subset)
    
train_dataset = concatenate_datasets(train_subset).shuffle(seed = 42)
    
```

```python
# Filtering 4k reviews from testing dataset for each class => 20k test dataset in total

test = dataset["test"].shuffle(seed = 42)

test_subset = []

for label in range(5):
    subset = test.filter(
        lambda x: x["label"] == label
    ).select(range(4000))
    
    test_subset.append(subset)
    
test_dataset = concatenate_datasets(test_subset).shuffle(seed = 42)
```

```python
# Verification of the distribution

from collections import Counter
#Counter is a class in collections module which counts how many times each unique value appears in a list or any iterable

print(Counter(train_dataset["label"]))
print(Counter(test_dataset["label"]))
```

```python
# Converting the datasets to pandas dataframe

train_df = train_dataset.to_pandas()
test_df = test_dataset.to_pandas()
```

```python
# Saving as csv file

train_df.to_csv("../data/raw/train.csv", index= False)
test_df.to_csv("../data/raw/test.csv", index = False)
```
